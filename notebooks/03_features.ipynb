```python
import sqlite3
import pandas as pd
from pathlib import Path

ROOT = Path.home() / "Documents" / "customer-churn-retention"

conn = sqlite3.connect(ROOT / "data" / "churn.db")

with open(ROOT / "sql" / "02_features.sql") as f:
    features = pd.read_sql(f.read(), conn)

print(features.shape)
features.to_csv(ROOT / "data" / "processed" / "features.csv", index=False)
features.head()
```

```python
features.groupby("tenure_group")["churn_value"].agg(["count", "mean"]).round(3)
```
